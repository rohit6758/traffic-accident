# Traffic Intelligence System
## Phase 1: Data Preparation & ML Pipeline

In this notebook, we will:
1. Load traffic data
2. Perform simple Exploratory Data Analysis (EDA)
3. Train a regression model to predict Traffic Volume
4. Train a classification model to predict Accident Risk
5. Save our models for the web app.

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score, confusion_matrix, classification_report
import joblib
```

### Step 1: Load Data
We load the dataset and take a quick look.

```python
# 1. Load data
# Why: Reads the CSV file so we can analyze the data.
df = pd.read_csv('../data/traffic_data.csv')
print("Data Shape:", df.shape)
df.head()
```

```python
# Check for missing values
# Why: Missing data can break machine learning models, so we check if there's any.
df.isnull().sum()
```

### Step 2: Exploratory Data Analysis
Let's see some simple plots to understand the data.

```python
# Plot Traffic Volume by Hour
# Why: Helps us visualize if there are rush hours.
plt.scatter(df['hour'], df['traffic_volume'], alpha=0.1)
plt.title('Traffic Volume vs Hour of Day')
plt.xlabel('Hour (0-23)')
plt.ylabel('Traffic Volume')
plt.show()
```

### Step 3: Traffic Volume Prediction (Regression)
We predict the continuous traffic volume based on conditions.

```python
# Select Features (X) and Target (y)
X = df[['hour', 'temperature', 'weather_code', 'is_weekend', 'road_quality']]
y_volume = df['traffic_volume']

# Split the data
# Why: We need some data to train the model and some to test it.
X_train_vol, X_test_vol, y_train_vol, y_test_vol = train_test_split(X, y_volume, test_size=0.2, random_state=42)
```

```python
# Train a Linear Regression model
# Why: It is a simple model to predict numerical values.
volume_model = LinearRegression()
volume_model.fit(X_train_vol, y_train_vol)
```

```python
# Predict and Evaluate
# Why: We check how well the model predicts on the test data.
y_pred_vol = volume_model.predict(X_test_vol)

print("R2 Score:", r2_score(y_test_vol, y_pred_vol))
print("Mean Squared Error:", mean_squared_error(y_test_vol, y_pred_vol))
```

### Step 4: Accident Risk Prediction (Classification)
We predict if the accident risk is Low (0) or High (1).

```python
y_risk = df['accident_risk']

# Split the data for classification
X_train_risk, X_test_risk, y_train_risk, y_test_risk = train_test_split(X, y_risk, test_size=0.2, random_state=42)
```

```python
# Train Logistic Regression model
# Why: A simple algorithm to classify binary outcomes (0 or 1).
risk_model = LogisticRegression()
risk_model.fit(X_train_risk, y_train_risk)
```

```python
# Evaluate the classification model
y_pred_risk = risk_model.predict(X_test_risk)

print("Accuracy:", accuracy_score(y_test_risk, y_pred_risk))
print("\nConfusion Matrix:\n", confusion_matrix(y_test_risk, y_pred_risk))
print("\nClassification Report:\n", classification_report(y_test_risk, y_pred_risk))
```

### Step 5: Save Models
We save the models so our app can use them.

```python
# Save the models
# Why: Saves the trained models to disk so we don't have to retrain them in the app.
joblib.dump(volume_model, '../models/volume_model.pkl')
joblib.dump(risk_model, '../models/risk_model.pkl')
print("Models saved successfully!")
```
